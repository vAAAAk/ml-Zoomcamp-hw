```python
df_asia = df[df.origin == "Asia"]
df_asia.head()
```

```python
df_asia = df_asia[["vehicle_weight","model_year"]][0:7]
```

```python
df_asia
```

```python
X = df_asia.values
X
```

```python
XTX = X.T.dot(X)
XTX
```

```python
XTX_inverse = np.linalg.inv(XTX)
XTX_inverse
```

```python
y = [1100, 1300, 800, 900, 1000, 1100, 1200]
```

```python
w = XTX_inverse.dot(X.T)
w
```
